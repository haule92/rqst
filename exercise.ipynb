```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    'date':['2025-02-28', '2025-03-31', '2025-04-17', '2025-04-30', '2025-05-31', '2025-06-30', '2025-09-15', '2025-09-30', '2025-10-31'],
    'stockA': [100, 101.3, 102.8, 101.9, 103.4, 105.3, 104.2, 101.5, 100.9],
})
```

```python
df['date'] = pd.to_datetime(df['date'])
df.set_index('date', inplace=True)
```

```python
ret = df / df.shift() -1
```

```python
mean_per_month = ret.resample('ME').mean()
rev_mask_notna = mean_per_month['stockA'].notna()[::-1]
rev_weights = np.maximum(0.5**(np.cumsum(rev_mask_notna)), 0.0625) * rev_mask_notna
rev_weights[rev_weights.cumsum() > 1] = 0
mean_per_month['weights'] = rev_weights[::-1]
mean_per_month['weighted_return'] = mean_per_month['stockA'] * mean_per_month['weights']
avg_ret = mean_per_month['weighted_return'].sum()
```
